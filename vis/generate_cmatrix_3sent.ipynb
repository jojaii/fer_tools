```python
import os
import numpy as np
from sklearn.metrics import confusion_matrix
```

```python
class Entry:
    def __init__(self, sentiment_probs, frame_number=None, fps=None):
        self.frame_number = frame_number
        self.fps = fps 
        self.sentiment_probs = sentiment_probs
```

```python
# TODO: entry class
# class attributes are: frame, fps, and sentiment_probs

# each filename is a label
directory = "/home/ubuntu/Desktop/Bravo_Analysis/AffectNet"
# FER2013 labels
labels = ["angry", "disgust", "fear"]#, "happy", "sad", "surprise", "neutral"]
files = ["angry.txt", "disgust.txt", "fear.txt"]#, "happy.txt", "sad.txt", "surprise.txt", "neutral.txt"]

# AffectNet labels
# labels = ["neutral", "happy", "sad", "surprise", "fear", "disgust", "angry"]
# # files = ["neutral.txt", "happy.txt", "sad.txt", "surprise.txt", "fear.txt", "disgust.txt", "angry.txt"]
# labels = ["fear", "disgust", "angry"]
# files = ["fear.txt", "disgust.txt", "angry.txt"]
```

```python
# 1. fix append

# initialize entry list
entry_list = []
# actual values in confusion matrix
y_actual = []
for file in files:
    open_file = open(os.path.join(directory,file), "r")

    lines = open_file.read().splitlines()
    frame_count = len(lines)

    # get index of current opened file
    correct_label = files.index(file)
#     if file == "fear.txt": correct_label = 4
#     elif file == "disgust.txt": correct_label = 5
#     elif file == "angry.txt": correct_label = 6
    
    # populate array with correct labels according to number of frames in the file
    # (represented by frame_count)
    # TODO: append([correct_label] * frame_count) then just flatten the array after
#     y_actual = [labels[correct_label]] * frame_count
    # for now, use 40 for max frame_count for uniformity
    frame_count = 500
    for i in range(0, frame_count):
        y_actual.append(correct_label)

    for line in lines:
        # only take first 40 frames
        if lines.index(line) < frame_count:
            temp = line.split(",")
            # TODO: how to separate different frame numbers of different files??
            frame_number = temp[0]
            fps = temp[1]
            # extract first to last sentiment probability
            sentiment_probs = np.array(temp[2:len(temp)])
            sentiment_probs = sentiment_probs.astype(np.float)
#             print(sentiment_probs)
            entry = Entry(sentiment_probs, frame_number, fps)
            entry_list.append(entry)
    open_file.close()
```
