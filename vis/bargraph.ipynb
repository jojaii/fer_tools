```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# TODO: entry class
# class attributes are: frame, fps, and sentiment_probs

file = "/home/ubuntu/Desktop/Bravo_Analysis/face_count_haar.txt"

open_file = open(file, "r")

frames = {}
lines = open_file.read().splitlines()
frame_count = len(lines)
for line in lines:
    temp = line.split(" ")
#     print(temp)
    # TODO: how to separate different frame numbers of different files??
    fps = float(temp[0])
    face_count = int(temp[1])
    if int(face_count) in frames:
        # append to list
        frames[face_count].append(fps)
    else:
#         print(face_count)
#         print(fps)
        frames[face_count] = [fps]

print(frames.keys())

y_axis = list(frames.keys())
x_axis = []
for item in frames.values():
    average_fps = sum(item)/(float(len(item)))
    x_axis.append(average_fps)

df = pd.DataFrame({'Detected faces':y_axis, 'Frames per second':x_axis})
ax = df.plot.bar(x='Detected faces', y='Frames per second', rot=0)

ax.plot()
plt.show()
```
