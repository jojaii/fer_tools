```python
input_size = [320,352,416,480,512,608]
mean_fps = [9.790, 8.840, 7.469, 6.217, 5.980, 4.644]
mean_tpr = [56.1, 64.0, 75.4, 81.2, 82.1, 85.9]
```

```python
plt.plot(input_size, mean_fps, 'k-o')
plt.tick_params(axis = 'both', labelsize = 14)
plt.xlabel('Input Image Size (square)', family = 'serif', fontsize = 14)
plt.ylabel('Average Frames Per Second', family = 'serif', fontsize = 14)
```

```python
plt.plot(input_size, mean_tpr, 'k-o')
plt.tick_params(axis = 'both', labelsize = 14)
plt.xlabel('Input Image Size', family = 'serif', fontsize = 14)
plt.ylabel('Average True Positive Rate', family = 'serif', fontsize = 14)
```

```python
inference_time = [61, 125, 172, 22, 29, 51, 125, 85]
mAP = [28, 31.2, 36.2, 28.2, 31.0, 33.0, 36, 29.9]
markers = ["x" , "," , "o" , "v" , "^" , "<", ">", "1"]
colors = ['violet','g','b','c','m', 'y', 'k', 'orange']
labels = ['SSD-321', 'SSD-513', 'FPN-FRCN', 'YOLOv3-320', 'YOLOv3-416', 'YOLOv3-608', 'RetinaNet-101-600', 'R-FCN']
```

```python


for i in range(len(inference_time)):
    plt.axvline(x = 100, color = 'r', linestyle = '-')
    plt.axhline(y = 32, color = 'r', linestyle = '-')
    
    plt.scatter(inference_time[i], mAP[i], 
                color = colors[i], marker = markers[i], s = 100, label = labels[i])
    
    plt.tick_params(axis = 'both', labelsize = 14)
    plt.xticks(np.arange(30, 220, step=30))
    
    plt.xlabel('Inference Time (in ms)', family = 'serif', fontsize = 14)
    plt.ylabel('Mean Average Precision (in %)', family = 'serif', fontsize = 14)
    plt.legend(loc='upper center', bbox_to_anchor=(0.5, -.20),
          ncol=4, fancybox=False, shadow=False)
```
